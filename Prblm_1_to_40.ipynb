```python
class Mysum:
  def __init__(self, l):
    self.l = l

  def __str__(self):
    return f"Sum of the list is : {sum(self.l)}"

a = Mysum([1,2,3,4,5])
print(a)
```

```python
from functools import reduce

class Mymult:
  def __init__(self, l):
    self.l = l

  def __str__(self):
    return f"Multi of the list is : {reduce(lambda x,y : x * y , self.l)}"

a = Mymult([1,2,3,4,5])
print(a)
```

```python
class Largest:
  def __init__(self, l):
    self.l = l

  def __str__(self):

    a = self.l[0]

    for i in self.l:
      if a < i:
        a = i

    return f"Largest number of the list is : {a}"

a = Largest([1,2,3,4,5])
print(a)
```

```python
class Smallest:
  def __init__(self, l):
    self.l = l

  def __str__(self):

    a = self.l[0]

    for i in self.l:
      if a > i:
        a = i

    return f"Smallest number of the list is : {a}"

a = Smallest([1,2,3,4,5])
print(a)
```

```python
class string:
  def __init__(self, s):
    self.s = s

  def __str__(self):
    n = 0

    for word in self.s:
      if (len(word) >= 2) and (word[0] == word[-1]):
        n += 1

    return f"{n}"

s = string(['abc', 'xyz', 'aba', '1221'])
print(s)
```

```python
class sorting:
  def __init__(self, l):
    self.l = l

  def last(self, sub):
    return sub[-1]

  def sorted_list(self):
    return sorted(self.l, key = self.last)

s = sorting([(2, 5), (1, 2), (4, 4), (2, 3), (2, 1)])
s.sorted_list()
```

```python
class sorting:
  def __init__(self, l):
    self.l = l

  def sorted_list(self):
    return sorted(self.l, key = lambda x : x[-1])

s = sorting([(2, 5), (1, 2), (4, 4), (2, 3), (2, 1)])
s.sorted_list()
```

```python
# 8
class empty:
  def __init__(self, l):
    self.l = l

  def __str__(self):
    if len(self.l) > 0:
      return f"Not Empty"
    else:
      return "Empty"

e = empty([3])
print(e)
```

```python
# 9

class copy:
  def init(self, l):
    self.l = l

  def copy_l(self):
    return self.l.copy() * 4

c = copy()
c.init([1,2,3,4,5])
print(c.copy_l())
```

```python
# 10
class find:
  def __init__(self, l, n):
    self.l = l
    self.n = n

  def find_word(self):
    a = []
    for word in self.l:
      if len(word) > self.n:
        a.append(word)

    return a


f = find(['masum', 'noman','shafi','kalami'], 5)
f.find_word()
```

```python
# 11
class common:
  def __init__(self, l1, l2):
    self.l1 = l1
    self.l2 = l2

  def find_common(self):
    for i in self.l1:
      if i in self.l2:
        return True
    else:
      return False

c = common([1,2,3,4,53],[5,6,7,8,9])
c.find_common()
```

```python
# 14
class remove_even:
  def __init__(self, l):
    self.l = l

  def remove(self):
    for i in range(0, len(self.l)):
      self.l = [i for i in self.l if i % 2 != 0]
    return self.l

r = remove_even([1,2,3,4,5,6,7,8])
r.remove()
```

```python
# 19
class difference:
  def __init__(self, l1, l2):
    self.l1 = l1
    self.l2 = l2

  def diff(self):
    d1 =  list(set(self.l1) - set(self.l2))
    d2 =  list(set(self.l2) - set(self.l1))

    return d1 + d2

d = difference([1,2,3,4,5],[4,5,6,7,8])
d.diff()
```
