```python
import pandas as pd
import numpy as np
```

```python
# can we have multiple index? Let's try
index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]

a = pd.Series([1,2,3,4,5,6,7,8], index = index_val)
a
# a[('cse', 2020)]
```

```python
# The problem?
# a['cse']
```

```python
# The solution -> multiindex series(also known as Hierarchical Indexing)
# multiple index levels within a single index
```

```python
# how to create multiindex object
# 1. pd.MultiIndex.from_tuples()
index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
Multiindex = pd.MultiIndex.from_tuples(index_val)
Multiindex

# 2. pd.MultiIndex.from_product()
pd.MultiIndex.from_product([['cse','ece'], [2019,2020,2021,2022]])
```

```python
# creating a series with multiindex object

s = pd.Series([1,2,3,4,5,6,7,8], index=Multiindex)
s
```

```python
# how to fetch items from such a series
s[('cse',2019)]
s['cse']
```

```python
# unstack
temp = s.unstack()
temp
```

```python
# stack
temp.stack()
```

```python
# Multiindex Data fream

branch_df1 = pd.DataFrame(
    [
        [1,2],
        [3,4],
        [5,6],
        [7,8],
        [9,12],
        [10,13],
        [11,15],
        [12,20]
    ],
    index = Multiindex,
    columns = ['avg_packge', 'students']
)
branch_df1
```

```python
# Multiindex DataFream

branch_df2 = pd.DataFrame(
    [
      [1,2,0,0],
      [3,4,0,0],
      [5,6,0,0],
      [7,8,0,0],
    ],

    index = [2019, 2020, 2021, 2022],
    columns = pd.MultiIndex.from_product([['delhi', 'mumbai'], ['avg_packge', 'students']])

)

branch_df2
```

```python
branch_df2.loc[2019]['delhi']
```

```python
# Multiindex series and DataFream

branch_df3 = pd.DataFrame(
    [
      [1,2,0,0],
      [3,4,0,0],
      [5,6,0,0],
      [7,8,0,0],
      [9,10,0,0],
      [11,12,0,0],
      [13,14,0,0],
      [15,16,0,0],
    ],

    index = Multiindex,
    columns = pd.MultiIndex.from_product([['delhi', 'mumbai'], ['avg_packge', 'students']])

)

branch_df3
```

```python
branch_df3.loc['cse']['delhi']['avg_packge'][2019]
```

# Stacking and Unstacking

```python
branch_df3.stack(future_stack=True).stack(future_stack=True)
```

```python
branch_df3.unstack().unstack()
```

```python
# Extracing rows single

branch_df3.loc[('cse', 2021)]
```

```python
# multiple rows

branch_df3.loc[('cse', 2019):('ece', 2020) : 2]
```

```python
# using iloc
branch_df3.iloc[0:5:2]
```

```python
# single col

branch_df3['delhi']['students']
```

```python
# multiple cols
branch_df3['delhi']
```

```python
branch_df3.iloc[:,0:3:2]
```

```python
# both
branch_df3.iloc[[1,4] , [1,2]]
```

```python
# sort index
branch_df3.sort_index(ascending = False)
branch_df3.sort_index(ascending = [False, True])  # level 0 decending level 1 ascending
branch_df3.sort_index(level = 1, ascending=False) # sort year index
```

```python
# Multiindex dataFrame -> transpose
branch_df3.transpose()
```

```python
# swaplevel

branch_df3.swaplevel()
branch_df3.swaplevel(axis = 1)
```

# Melt and pivet

```python
# Melt

deaths = pd.read_csv("/content/time_series_covid19_deaths_global.csv")
confirm = pd.read_csv("/content/time_series_covid19_confirmed_global.csv")
```

```python
deaths.head()
```

```python
confirm
```

```python
deaths = deaths.melt(id_vars = ['Province/State',	'Country/Region',	'Lat',	'Long'], var_name='Date', value_name='num_death')
```

```python
confirm = confirm.melt(id_vars = ['Province/State',	'Country/Region',	'Lat',	'Long'], var_name='Date', value_name='num_case')
```

```python
confirm.merge(deaths, on = ['Province/State',	'Country/Region',	'Lat','Long','Date'])[['Country/Region', 'Date',	'num_case',	'num_death']]
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
df.groupby('sex', observed=False)[['total_bill']].mean()
```

```python
df.groupby(['sex', 'smoker'], observed=False)[['total_bill']].mean().unstack()
```

```python
df.pivot_table(index = 'sex', columns = 'smoker', values = 'total_bill', observed=False)
```

```python
# aggfunc

pd.pivot_table(df, index = 'sex', columns = 'smoker', values = 'total_bill', aggfunc = 'sum', observed=False)
```

```python
pd.pivot_table(df, index = 'sex', columns = 'smoker', aggfunc = 'sum', observed=True)
```

```python
# multidimensional

pd.pivot_table(df, index = ['sex', 'smoker'], columns=['day', 'time'], aggfunc={'size':'mean', 'tip' : 'max', 'total_bill' : 'sum'}, margins=True, observed=False)
```

```python
# # margins

pd.pivot_table(df, index = 'sex', columns = 'smoker', values = 'total_bill', margins=True, observed= True)
```
