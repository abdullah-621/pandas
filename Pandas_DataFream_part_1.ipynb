```python
def against(Team1, Team2):
    # দুইভাবে ম্যাচ ধরার জন্য condition বানাচ্ছি
    matches = ipl[((ipl['Team1'] == Team1) & (ipl['Team2'] == Team2)) |
                  ((ipl['Team1'] == Team2) & (ipl['Team2'] == Team1))]

    total_matches = matches.shape[0]
    team1_win = matches[matches['WinningTeam'] == Team1].shape[0]
    team2_win = matches[matches['WinningTeam'] == Team2].shape[0]

    return (f"Total Matches between {Team1} and {Team2}: {total_matches}\n"
            f"{Team1} Wins: {team1_win}\n"
            f"{Team2} Wins: {team2_win}")

# Example
result = against('Royal Challengers Bangalore', 'Chennai Super Kings')
print(result)
```
