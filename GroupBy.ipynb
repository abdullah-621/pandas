```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv("imdb-top-1000.csv")
```

```python
geners = movies.groupby('Genre')
```

```python
# find the top 3 geners by total earning
movies.groupby('Genre')['Gross'].sum(numeric_only=True).sort_values(ascending = False).head(3)
```

```python
# Find the Genre with highest avg IMDB rating

movies.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

```python
# find the most popular director
movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
# find the highest rated movie of each genre
movies.groupby('Genre')['IMDB_Rating'].max()
```

```python
# agg method
# passing dict

geners.agg(
    {
        'Runtime' : 'mean',
        'IMDB_Rating' : 'mean',
        'No_of_Votes' : 'sum',
        'Gross' : 'max',
        'Metascore' : 'min',
        'Series_Title' : 'sum'
    }
)
```

```python
# passing list

geners.agg({
        'Runtime': ['mean', 'min', 'max'],
        'IMDB_Rating': ['mean', 'min', 'max'],
        'No_of_Votes': ['mean'],
        'Gross': 'max',
        'Metascore': ['mean']
      })
```

```python
df = pd.DataFrame(columns=movies.columns)

for group,data in geners:
  df = df.append(data[data['IMDB_Rating'] == data['IMDB_Rating'].max()])

df
```

```python

dfs = []  # এখানে সব group-এর data জমা হবে

for group, data in geners:
    dfs.append(data[data['IMDB_Rating'] == data['IMDB_Rating'].max()])

# সব group-এর top-rated movie combine করা
# df = pd.concat(dfs, ignore_index=True)

# df

dfs
```

```python
# split apply combination
# apply -> builtin function

geners.apply(min)
geners.apply(np.minimum.reduce)
```

```python
# find number of movies starting with A for each group

def foo(series):
  return (series.str.startswith('A').sum())
  # print(type(series))

geners['Series_Title'].apply(foo)
```

```python
# find ranking of each movie in the group according to IMDB score

def ranking(group):
  #  group = group.copy()
   group['Ranking'] = group['IMDB_Rating'].rank(ascending = False)
   return group


geners.apply(ranking)
```

```python
# find normalized IMDB Rating group wise

def normalized(group):
  group['Normal_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min()) / (group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
  return group

geners.apply(normalized)
```
