###`Q-6:` Make a dataframe of each team in IPL with details like - Team Name, Matches Played, Win%, Home Win%, Away Win%.
Show sorted dataframe on Win%

Replace old team name as new name before performing any tasks.
```
Delhi Daredevils ->Delhi Capitals
Kings XI Punjab -> Punjab Kings
Rising Pune Supergiants -> Rising Pune Supergiant
```

Note: Team1 represents Home team. Exclude not result matches.


````python
ipl = pd.read_csv('ipl-matches.csv')
````

````python
team_home = {
    'Chennai Super Kings': 'Chennai',
    'Deccan Chargers': 'Hyderabad',
    'Delhi Capitals': 'Delhi',
    'Delhi Daredevils': 'Delhi',   # পুরোনো নাম
    'Gujarat Lions': 'Rajkot',
    'Gujarat Titans': 'Ahmedabad',
    'Kings XI Punjab': 'Chandigarh',
    'Punjab Kings': 'Chandigarh',  # নতুন নাম
    'Kochi Tuskers Kerala': 'Kochi',
    'Kolkata Knight Riders': 'Kolkata',
    'Lucknow Super Giants': 'Lucknow',  # নতুন দল
    'Mumbai Indians': 'Mumbai',
    'Pune Warriors': 'Pune',
    'Rajasthan Royals': 'Jaipur',
    'Rising Pune Supergiant': 'Pune',
    'Rising Pune Supergiants': 'Pune',  # spelling variation
    'Royal Challengers Bangalore': 'Bangalore',
    'Sunrisers Hyderabad': 'Hyderabad'
}

def matches_played(team):
  new_df = ipl[(ipl['Team1'] == team) | (ipl["Team2"] == team)]
  return new_df.shape[0]

def total_wins(team):
  # Win = ((ipl['WinningTeam'] == team).sum() / matches_played(team)) * 100
  return (ipl['WinningTeam'] == team).sum()


def home_wins(team):
    city = team_home[team]
    return ((ipl['WinningTeam'] == team) & (ipl['City'] == city)).sum()

def away_wins(team):
    city = team_home[team]
    return ((ipl['WinningTeam'] == team) & (ipl['City'] != city)).sum()



  # code here
teams = np.unique(np.concatenate((ipl['Team1'].unique(), ipl['Team2'].unique())))

data = {
    'Team Name' : [],
    'Matches Played' : [],
    'Wins%' : [],
    'Home Win%' : [],
    'Away Win%' : []
}

for team in teams:
  total = matches_played(team)
  wins = total_wins(team)
  home = home_wins(team)
  away = away_wins(team)

  win_perc = round(wins / total * 100, 2)
  home_prec = round(home / total * 100, 2)
  away_prec = round(away / total * 100, 2)

  data['Team Name'].append(team)
  data['Matches Played'].append(total)
  data['Wins%'].append(win_perc)
  data['Home Win%'].append(home_prec)
  data['Away Win%'].append(away_prec)


new_df = pd.DataFrame(data)
new_df.sort_values(by='Wins%', ascending=False)


````

###`Q-9:` IPL Point Table

Make a function `point_table` which take `season` as parameter and show points table in non-ascendng order of points and in ascending order of team name.

For winning - 2 Ponits;
For loosing - 0 Point
For not result both team gets 1 points.

Make dataframe which will have
`TeamName` `MatchesPlayed` `MatchesWon` `NoResult` `Points`
make `TeamName` as index.

```
season parametr should be one of these->
['2022', '2021', '2020/21', '2019', '2018', '2017', '2016', '2015', '2014', '2013', '2012', '2011', '2009/10', '2009', '2007/08']
```


Output of two Top 2 in season 2022
```
TeamName    MatchesPlayed	MatchesWon	NoResult	Points

Gujarat Titans	    16	   12	       0	     24
Rajasthan Royals	  17	   10	       0	     20

```

````python
# code here
def MatchesPlayed(team, this_season):
  return ((this_season['Team1'] == team) | (this_season['Team2'] == team)).sum()

def MatchesWon(team, this_season):
  return (this_season['WinningTeam'] == team).sum()


def point_table(season):
  this_season = ipl[ipl['Season'] == season]

  teams = np.unique(np.concatenate((this_season['Team1'].unique(), this_season['Team2'].unique())))

  table = []

  for team in teams:
    Match_paly = MatchesPlayed(team, this_season)
    Match_win = MatchesWon(team, this_season)

    table.append([team, Match_paly, Match_win, Match_win*2])

  new_table = pd.DataFrame(table, columns=['TeamName', 'MatchesPlayed','MatchesWon','Points'])
  return new_table.sort_values(by = ['Points','TeamName'], ascending=[False,True])

table = point_table('2022')
table
````
