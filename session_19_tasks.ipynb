### `Q-4:` Perform `groupby` based on the `"Team"` column and then perform Z Normalization on top of the below columns of each group:
1. Passes
2. Passes Completed
3. Attempted Line Breaks
4. Completed Line Breaks

You have to make a python function named `z_normalization` which takes two arguments:

1. *group:* Every group that you have created
2. *cols_to_perform:* This parameter takes a list of columns on which you have to perform the Z-Normalization.

For this problem, you have to use th `apply()` method.

$\Large Z \ - \ Normalization = \frac{X_i - \mu}{std} $

After that find out the below values for each group:
- minimum "Passess"
- maximum "Passess"
- minimum "Yellow Cards"
- maximum "Yellow cards"
- average "Yellow Cards"
- maximum "Attempted Line Breaks"
- minimum "Attempted Line Breaks"
- standardard deviation of "Attempted Line Breaks"
- average Possession

````python
# code here
def nomalized(group):
  group['Passes_normal'] = (group['Passes'] - group['Passes'].min()) / (group['Passes'].max() - group['Passes'].min())
  group['Passes Completed_normal'] = (group['Passes Completed'] - group['Passes Completed'].min()) / (group['Passes Completed'].max() - group['Passes Completed'].min())
  group['Attempted Line Breaks_normal'] = (group['Attempted Line Breaks'] - group['Attempted Line Breaks'].min()) / (group['Attempted Line Breaks'].max() - group['Attempted Line Breaks'].min())
  group['Completed Line Breaks_normal'] = (group['Completed Line Breaks'] - group['Completed Line Breaks'].min()) / (group['Completed Line Breaks'].max() - group['Completed Line Breaks'].min())

  return group

temp_df = fifa_df.groupby("Team").apply(nomalized)

result = temp_df.agg(
    {
        "Passes" : ['min', 'max'],
        "Yellow Cards" : ['min', 'max', 'mean'],
        "Attempted Line Breaks" : ['max', 'min', 'std'],
        "Possession (%)" : 'mean'
    }
).round(2)

result
````

###`Q-8:` Make a dataframe for all batting pairs played together.
```
Batsman1 Batsman2 Runs Avg StrikeRate
```

> Just to ease this question you can count wide-balls for strike rate.

````python
# code here
dfs = {
    'Batsman1' :[],
    'Batsman2' : [],
    'Runs' : [],
    'Avg' : [],
    'StrikeRate' : []
}

batting_duo = ipl.groupby(['batter', 'non-striker'])

for (b1,b2), data in batting_duo:

  runs = data['total_run'].sum()

  dfs['Batsman1'].append(b1)
  dfs['Batsman2'].append(b2)
  dfs['Runs'].append(runs)
  dfs['Avg'].append((runs / data.shape[0]).round(2))
  dfs['StrikeRate'].append((data['extra_type'] == 'wides').sum())

new_df = pd.DataFrame(dfs).sort_values(by = 'Runs', ascending=False).reset_index(drop=True)
new_df.head(10)

````

````python
# code here
dfs = {
    'Batsman1' :[],
    'Batsman2' : [],
    'Runs' : [],
    'Avg' : [],
    'StrikeRate' : []
}

batting_duo = ipl.groupby(['batter', 'non-striker'])

for (b1,b2), data in batting_duo:

  runs = data['total_run'].sum()
  balls = len(data) - (data['extra_type'] == 'wides').sum() # without wide ball
  strike_rate = (runs / balls * 100).round(2) if balls > 0 else 0

  dfs['Batsman1'].append(b1)
  dfs['Batsman2'].append(b2)
  dfs['Runs'].append(runs)
  dfs['Avg'].append((runs / data.shape[0]).round(2))
  dfs['StrikeRate'].append(strike_rate)

new_df = pd.DataFrame(dfs).sort_values(by = 'Runs', ascending=False).reset_index(drop=True)
new_df.tail(10)

````
