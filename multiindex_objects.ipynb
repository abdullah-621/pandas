```python
import pandas as pd
import numpy as np
```

```python
# can we have multiple index? Let's try
index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]

a = pd.Series([1,2,3,4,5,6,7,8], index = index_val)
a
# a[('cse', 2020)]
```

```python
# The problem?
a['cse']
```

```python
# The solution -> multiindex series(also known as Hierarchical Indexing)
# multiple index levels within a single index
```

```python
# how to create multiindex object
# 1. pd.MultiIndex.from_tuples()
index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
Multiindex = pd.MultiIndex.from_tuples(index_val)
Multiindex

# 2. pd.MultiIndex.from_product()
pd.MultiIndex.from_product([['cse','ece'], [2019,2020,2021,2022]])
```

```python
# creating a series with multiindex object

s = pd.Series([1,2,3,4,5,6,7,8], index=Multiindex)
s
```

```python
# how to fetch items from such a series
s[('cse',2019)]
s['cse']
```

```python
# unstack
temp = s.unstack()
temp
```

```python
# stack
temp.stack()
```

```python
# Multiindex Data fream

branch_df1 = pd.DataFrame(
    [
        [1,2],
        [3,4],
        [5,6],
        [7,8],
        [9,12],
        [10,13],
        [11,15],
        [12,20]
    ],
    index = Multiindex,
    columns = ['avg_packge', 'students']
)
branch_df1
```

```python
# Multiindex DataFream

branch_df2 = pd.DataFrame(
    [
      [1,2,0,0],
      [3,4,0,0],
      [5,6,0,0],
      [7,8,0,0],
    ],

    index = [2019, 2020, 2021, 2022],
    columns = pd.MultiIndex.from_product([['delhi', 'mumbai'], ['avg_packge', 'students']])

)

branch_df2
```

```python
branch_df2.loc[2019]['delhi']
```

```python
# Multiindex series and DataFream

branch_df3 = pd.DataFrame(
    [
      [1,2,0,0],
      [3,4,0,0],
      [5,6,0,0],
      [7,8,0,0],
      [9,10,0,0],
      [11,12,0,0],
      [13,14,0,0],
      [15,16,0,0],
    ],

    index = Multiindex,
    columns = pd.MultiIndex.from_product([['delhi', 'mumbai'], ['avg_packge', 'students']])

)

branch_df3
```

```python
branch_df3.loc['cse']['delhi']['avg_packge'][2019]
```

# Stacking and Unstacking
