```python
class Mysum:
  def __init__(self, l):
    self.l = l

  def __str__(self):
    return f"Sum of the list is : {sum(self.l)}"

a = Mysum([1,2,3,4,5])
print(a)
```

```python
from functools import reduce

class Mymult:
  def __init__(self, l):
    self.l = l

  def __str__(self):
    return f"Multi of the list is : {reduce(lambda x,y : x * y , self.l)}"

a = Mymult([1,2,3,4,5])
print(a)
```

```python
class Largest:
  def __init__(self, l):
    self.l = l

  def __str__(self):

    a = self.l[0]

    for i in self.l:
      if a < i:
        a = i

    return f"Largest number of the list is : {a}"

a = Largest([1,2,3,4,5])
print(a)
```

```python
class Smallest:
  def __init__(self, l):
    self.l = l

  def __str__(self):

    a = self.l[0]

    for i in self.l:
      if a > i:
        a = i

    return f"Smallest number of the list is : {a}"

a = Smallest([1,2,3,4,5])
print(a)
```

```python
class string:
  def __init__(self, s):
    self.s = s

  def __str__(self):
    n = 0

    for word in self.s:
      if (len(word) >= 2) and (word[0] == word[-1]):
        n += 1

    return f"{n}"

s = string(['abc', 'xyz', 'aba', '1221'])
print(s)
```
