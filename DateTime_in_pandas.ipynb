```python
import pandas as pd
import numpy as np
```

# Date Time Object

```python
print(pd.Timestamp('2020/8/21'))
print(pd.Timestamp('2020-8-21'))
print(pd.Timestamp('2020, 8, 21'))
type(pd.Timestamp('5th january 2025'))
```

```python
# providing time also
print(pd.Timestamp('5th january 2025 9:34 PM'))
```

```python
# using datetime.datetime object

import datetime as dt

x = pd.Timestamp(dt.datetime(2025, 10, 26, 8, 25, 40))
x
```

```python
# fetching Attribute
x.year
x.month
x.day
x.hour
x.minute
x.second
```

# DatetimeIndex Object

```python
# from string

x = pd.DatetimeIndex(['2020/2/2', '2022/2/2', '2024/2/2'])
type(x)
```

```python
# using pyhton datetime objects

pd.DatetimeIndex([dt.datetime(2020,2,2), dt.datetime(2022,2,2), dt.datetime(2025,4,4)])
```

```python
# using pd.timestamp

df_index = pd.DatetimeIndex([pd.Timestamp(2020,2,2), pd.Timestamp(2022,2,2), pd.Timestamp(2025,4,4)])
```

```python
# using datetimeindex as a series index
pd.Series([1,2,3], index = df_index)
```
