# DatetimeIndex Object

```python
# from string

x = pd.DatetimeIndex(['2020/2/2', '2022/2/2', '2024/2/2'])
x
```

```python
# using pyhton datetime objects

pd.DatetimeIndex([dt.datetime(2020,2,2), dt.datetime(2022,2,2), dt.datetime(2025,4,4)])
```

```python
# using pd.timestamp

df_index = pd.DatetimeIndex([pd.Timestamp(2020,2,2), pd.Timestamp(2022,2,2), pd.Timestamp(2025,4,4)])
```

```python
# using datetimeindex as a series index
pd.Series([1,2,3], index = df_index)
```

# date_range function

```python
# genarate daily dates in a given range

pd.date_range(start = "2025/3/4", end = "2025/4/4", freq='2D')
```

```python
# Business day
pd.date_range(start = "2025/3/4", end = "2025/4/4", freq='B')
```

```python
# W -> One weak per day
pd.date_range(start = "2025/3/4", end = "2025/4/4", freq='W-THU')
```

```python
# h -> Hourly date(factor)

pd.date_range(start = "2025/3/4", end = "2025/4/4", freq='h')
```

```python
# ME -> Month end
pd.date_range(start = "2025/3/4", end = "2025/4/4", freq='ME')
```

```python
# MS -> Month start
pd.date_range(start = "2025/3/4", end = "2025/4/4", freq='MS')
```

```python
# YE -> year end
pd.date_range(start = "2025/3/4", end = "2030/4/4", freq='YE')
```

```python
# YS -> year start
pd.date_range(start = "2025/3/4", end = "2030/4/4", freq='YS')
```

```python
# using period
pd.date_range(start = "2025/3/3", periods = 25, freq = 'ME')
```

# to_datetime function

```python
# simple series example

s = pd.Series(['2023/4/5', '2034/4/5', '2056/6/7'])
# s.str.split('/').str.get(0)
pd.to_datetime(s).dt.month_name()
```

```python
# with error
s = pd.Series(['2023/4/5', '2034/4/5', '2056/69/7'])
pd.to_datetime(s, errors = 'coerce')
```
