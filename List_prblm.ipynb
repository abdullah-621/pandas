```python
# 18. Write a Python program to generate all permutations of a list in Python.

import itertools

l = [1,2,3]

for p in itertools.permutations(l):
  print(p)
```

```python
# 19. Write a Python program to get the difference between the two lists.

def difference(list1 , list2):
  list3 = []
  for i in list1:
    if i not in list2:
      list3.append(i)
  return list3


list1 = [1,2,3,4,5,6]
list2 = [4,5,6,7,8,9]

print("Difference between list1 to list2 : ", difference(list1, list2))
print("Difference between list2 to list1 : ", difference(list2, list1))




```

```python
# 22. Write a Python program to find the index of an item in a specified list.

l = [1,2,3,4,5,6,7]
k = int(input(f"Enter number from {l} : "))

for i in range(len(l)):
  if l[i] == k:
    print(f"Index of {k} is : {i}")
    break
```

```python
# 26. Write a python program to check whether two lists are circularly identical.

def isIdentical(list1, list2):

  if len(list1) != len(list2):
    return False

  for i in range(len(list1)):
    if list1[i] != list2[i]:
      return False

  return True



# A = [1, 2, 3, 4, 5]
# B = [3, 4, 5, 1, 2]

A = [1, 2, 3]
B = [2, 1, 3]

if isIdentical(A,B):
  print("Yes A and B is Identical")

else:
  for i in range(len(A)):

    A.append(A[0])
    A.pop(0)

    if isIdentical(A,B):
      print("Yes A and B is Identical")
      print(A)
      print(B)
      break
  else:
    print("No A and B are not Identical")
    print(A)
    print(B)
```
