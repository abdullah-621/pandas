```python
# 19. Write a Python program to get the difference between the two lists.

def difference(list1 , list2):
  list3 = []
  for i in list1:
    if i not in list2:
      list3.append(i)
  return list3


list1 = [1,2,3,4,5,6]
list2 = [4,5,6,7,8,9]

print("Difference between list1 to list2 : ", difference(list1, list2))
print("Difference between list2 to list1 : ", difference(list2, list1))




```

```python
# 26. Write a python program to check whether two lists are circularly identical.

def isIdentical(list1, list2):

  if len(list1) != len(list2):
    return False

  for i in range(len(list1)):
    if list1[i] != list2[i]:
      return False

  return True



# A = [1, 2, 3, 4, 5]
# B = [3, 4, 5, 1, 2]

A = [1, 2, 3]
B = [2, 1, 3]

if isIdentical(A,B):
  print("Yes A and B is Identical")

else:
  for i in range(len(A)):

    A.append(A[0])
    A.pop(0)

    if isIdentical(A,B):
      print("Yes A and B is Identical")
      print(A)
      print(B)
      break
  else:
    print("No A and B are not Identical")
    print(A)
    print(B)
```

```python
#34. Write a Python program using Sieve of Eratosthenes method for computing
# primes upto a specified number.
import math

def Sieve_of_Eratosthenes(n):
  primes = [0 for i in range(n + 1)]

  for j in range(2, int(math.sqrt(n)) + 1):
    for k in range(j ** j, n + 1, j):
      if k % j == 0:
        primes[k] = 1

  for l in range(2,len(primes)):
    if primes[l] != 1:
      print(l, end=" ")


n = int(input("Enter range : "))
print(f"Prime number of {n} range : ", end="")
Sieve_of_Eratosthenes(n)

```

```python
# 38. Write a Python program to change the position of every n-th value with the
# (n+1)th in a list.
l = [0,1,2,3,4,5]

for i in range(len(l) - 1):
  if i % 2 == 0:
    temp = l[i]
    l[i] = l[i + 1]
    l[i + 1] = temp
  else:
    continue


print(l)

```
